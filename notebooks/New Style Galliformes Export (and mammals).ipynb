```python
from pathlib import Path

import numpy as np
import pandas as pd
```

# galliformes

```python
galliformes_dir = Path("/data-ssd/alex/datasets/vision-export-20221020154449-Galliformes-573/")
galliformes_tax = pd.read_csv(galliformes_dir / "taxonomy.csv")
galliformes_leaf_tax = galliformes_tax[~galliformes_tax.leaf_class_id.isna()]
galliformes_leaf_taxon_ids = galliformes_leaf_tax.taxon_id.values
```

```python
ods_dir = Path("/data-ssd/alex/datasets/inat-open-data/inaturalist-open-data-20240827")
ods_obs_galliformes = pd.read_csv(ods_dir / "obs_galliformes.csv")
```

```python
ods_obs_galliformes.sample(5)
```

```python
ods_obs_galliformes.taxon_id.value_counts()
```

```python
!ls {ods_dir}
```

```python
ods_photos = pd.read_csv(
    ods_dir / "photos.csv",
    sep="\t",
)
```

```python
ods_obs_photos_galliformes = pd.merge(
    ods_obs_galliformes,
    ods_photos,
    left_on="observation_uuid",
    right_on="observation_uuid",
    how="left",
)
```

```python
ods_obs_photos_galliformes.to_csv(
    ods_dir / "obs_photos_galliformes.csv",
    index=False,
)
```

# now let's do mammals

```python
mammals_dir = Path("/data-ssd/alex/datasets/vision-export-20230412214019-Mammalia-40151/")
mammals_tax = pd.read_csv(mammals_dir / "taxonomy.csv")
mammals_leaf_tax = mammals_tax[~mammals_tax.leaf_class_id.isna()]
mammals_leaf_taxon_ids = set(mammals_leaf_tax.taxon_id.values)
```

```python
ods_dir = Path("/data-ssd/alex/datasets/inat-open-data/inaturalist-open-data-20240827")
ods_obs = pd.read_csv(
    ods_dir / "observations.csv",
    sep="\t",
    usecols=["observation_uuid", "taxon_id"],
)
```

```python
ods_obs.dropna(inplace=True)
```

```python
ods_obs.taxon_id = ods_obs.taxon_id.astype(np.int32)
```

```python
ods_obs.sample(5)
```

```python
mammal_ods_obs = ods_obs[ods_obs.taxon_id.isin(mammals_leaf_taxon_ids)]
len(mammal_ods_obs), len(ods_obs)
```

```python
mammal_ods_obs.to_csv(
    ods_dir / "obs_mammals.csv",
    index=False,
)
# mammal_ods_obs = pd.read_csv(ods_dir / "obs_mammals.csv")
```

```python
ods_photos = pd.read_csv(
    ods_dir / "photos.csv",
    sep="\t",
    usecols=["photo_id", "observation_uuid", "extension"],
)
```

```python
ods_obs_photos_mammals = pd.merge(
    mammal_ods_obs,
    ods_photos,
    left_on="observation_uuid",
    right_on="observation_uuid",
    how="left",
)
```

```python
ods_obs_photos_mammals.to_csv(
    ods_dir / "obs_photos_mammals.csv",
    index=False,
)
```
